```python
from sqlalchemy import create_engine, text, ForeignKey, UniqueConstraint
from sqlalchemy.orm import DeclarativeBase, mapped_column, Mapped, Session
from sqlalchemy import Integer, String, Boolean
from datetime import time
import os, json, copy
```

```python
os.makedirs("data", exist_ok=True)
engine = create_engine("sqlite:///data/museum.db")
```

```python
class Base(DeclarativeBase):
    pass
```

```python
class Museum(Base):
    __tablename__ = "museum"

    id: Mapped[int] = mapped_column(primary_key=True)
    museum_name: Mapped[str]
    url: Mapped[str]
    visited: Mapped[bool] = mapped_column(default=False)
```

```python
class Schedule(Base):
    __tablename__ = "schedule"

    id: Mapped[int] = mapped_column(primary_key=True)
    museum_id: Mapped[int] = mapped_column(ForeignKey("museum.id"))
    day_of_week: Mapped[int]
    open_time: Mapped[time]
    close_time: Mapped[time]
    last_entry_time: Mapped[time | None]
    is_closed: Mapped[bool] = mapped_column(default=False)

    __table_args__ = (UniqueConstraint("museum_id", "day_of_week"),)
```

```python
Base.metadata.create_all(engine)
```

```python
with engine.connect() as conn:
    conn.execute(text("DELETE FROM schedule"))
    conn.execute(text("DELETE FROM museum"))
    conn.commit()
```
